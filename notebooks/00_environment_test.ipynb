## Lakehouse Integration Test: Spark â†” Kafka â†” MinIO â†” ClickHouse

This notebook validates that the entire graduation project environment is successfully configured and communicating. We will perform the following steps:
1. **Initialize Spark Session** and verify Hadoop S3A, Iceberg, Kafka, and ClickHouse JDBC packages.
2. **MinIO Connection Test**: Write a dummy DataFrame to the `bronze` bucket.
3. **Kafka Connection Test**: Publish simulated sensor readings directly to Kafka.
4. **Bronze Layer Ingestion**: Read the simulated data from Kafka and write it to the `bronze` bucket as Parquet.
5. **Silver Layer Cleaning**: Read from the `bronze` bucket, cast data types, filter invalid rows, and write to the `silver` bucket.
6. **Gold Layer Aggregation**: Aggregate the cleaned readings to hourly averages and write to the `gold` bucket.
7. **Serving Layer Load**: Write the Gold aggregated metrics to the **ClickHouse** analytical database using the ClickHouse JDBC driver.
8. **Verify Serving Layer**: Run an SQL query on ClickHouse to verify that data is visible for Power BI/Grafana dashboard ingestion.

```python
import os
import time
from datetime import datetime
from pyspark.sql import SparkSession
from pyspark.sql.types import *
from pyspark.sql.functions import *

print("Initializing PySpark Session with pre-configured Lakehouse environment...")

# S3A & catalog configurations are pre-defined in spark-defaults.conf, we get or create the session
spark = SparkSession.builder.getOrCreate()

print(f"âœ… Spark Session created successfully! Spark version: {spark.version}")
```

```python
print("Step 1: Testing MinIO Object Storage connectivity...")

# Create test data
test_data = [("minio_check", "Success", datetime.utcnow())]
test_schema = StructType([
    StructField("check_name", StringType(), True),
    StructField("status", StringType(), True),
    StructField("timestamp", TimestampType(), True)
])

df_test = spark.createDataFrame(test_data, schema=test_schema)

# Write to MinIO bronze bucket
minio_path = "s3a://bronze/connection_test"
print(f"Writing test DataFrame to MinIO: {minio_path}")
df_test.write.mode("overwrite").parquet(minio_path)

# Read back from MinIO
df_read = spark.read.parquet(minio_path)
df_read.show()

print("âœ… MinIO connection verified! Read/Write works on 'bronze' bucket.")
```

```python
print("Step 2: Testing Kafka Broker connectivity and publishing mock sensor events...")

# Create mock raw sensor reading events
mock_events = [
    {"sensor_id": "FIELD-01-SM-01", "sensor_type": "soil_moisture", "reading_value": 34.2, "timestamp": "2026-06-26 10:05:00"},
    {"sensor_id": "FIELD-01-SM-02", "sensor_type": "soil_moisture", "reading_value": 41.8, "timestamp": "2026-06-26 10:06:00"},
    {"sensor_id": "FIELD-02-SM-01", "sensor_type": "soil_moisture", "reading_value": 29.5, "timestamp": "2026-06-26 10:10:00"},
    {"sensor_id": "FIELD-02-TMP-01", "sensor_type": "temperature", "reading_value": 28.3, "timestamp": "2026-06-26 10:10:00"},
    {"sensor_id": "FIELD-03-SM-01", "sensor_type": "soil_moisture", "reading_value": 52.1, "timestamp": "2026-06-26 10:12:00"}
]

# Define Schema
sensor_schema = StructType([
    StructField("sensor_id", StringType(), True),
    StructField("sensor_type", StringType(), True),
    StructField("reading_value", DoubleType(), True),
    StructField("timestamp", StringType(), True)
])

# Create Spark DataFrame from Python Dicts
df_mock = spark.createDataFrame(mock_events, schema=sensor_schema)

# Write to Kafka topic 'agri.sensors.raw'
# Kafka source requires 'key' and 'value' columns
df_kafka_payload = df_mock.selectExpr(
    "sensor_id as key",
    "to_json(struct(*)) as value"
)

print("Publishing messages to Kafka bootstrap servers at kafka:9092...")
df_kafka_payload.write \
    .format("kafka") \
    .option("kafka.bootstrap.servers", "kafka:9092") \
    .option("topic", "agri.sensors.raw") \
    .save()

print("âœ… Kafka publishing verified! Mock sensor data sent to topic 'agri.sensors.raw'.")
```

```python
print("Step 3: Simulating Ingestion to Bronze Layer (Batch load from Kafka)...")

# Load raw events from Kafka as a batch
df_kafka_raw = spark.read \
    .format("kafka") \
    .option("kafka.bootstrap.servers", "kafka:9092") \
    .option("subscribe", "agri.sensors.raw") \
    .option("startingOffsets", "earliest") \
    .load()

# Deserialize JSON payload
df_parsed = df_kafka_raw.select(
    col("key").cast("string").alias("kafka_key"),
    from_json(col("value").cast("string"), sensor_schema).alias("parsed_value"),
    col("partition").alias("kafka_partition"),
    col("offset").alias("kafka_offset")
).select("kafka_partition", "kafka_offset", "parsed_value.*")

# Write parsed raw data to MinIO Bronze layer
bronze_path = "s3a://bronze/sensor_readings"
print(f"Writing parsed data to Bronze Layer: {bronze_path}")
df_parsed.write.mode("overwrite").parquet(bronze_path)

# Verify Bronze contents
df_bronze_check = spark.read.parquet(bronze_path)
print(f"Bronze Layer Record Count: {df_bronze_check.count()}")
df_bronze_check.show(5, truncate=False)
```

```python
print("Step 4: Simulating Silver Layer ETL (Cleaning, Schema Enforcement & Quality Rules)...")

# Read from Bronze
df_bronze = spark.read.parquet("s3a://bronze/sensor_readings")

# Apply cleaning operations:
# 1. Cast timestamp to TimestampType
# 2. Filter out null values or extreme readings (outliers)
# 3. Add an ingestion audit timestamp
df_cleaned = df_bronze \
    .withColumn("timestamp", to_timestamp(col("timestamp"), "yyyy-MM-dd HH:mm:ss")) \
    .withColumn("reading_value", col("reading_value").cast(DoubleType())) \
    .withColumn("ingested_at", current_timestamp()) \
    .filter(col("reading_value").between(-10.0, 80.0)) # typical range for moisture / temperature

# Write to MinIO Silver layer
silver_path = "s3a://silver/sensor_readings"
print(f"Writing cleaned data to Silver Layer: {silver_path}")
df_cleaned.write.mode("overwrite").parquet(silver_path)

# Verify Silver contents
df_silver_check = spark.read.parquet(silver_path)
print(f"Silver Layer Record Count: {df_silver_check.count()}")
df_silver_check.show(5, truncate=False)
```

```python
print("Step 5: Simulating Gold Layer Business Transformations (Hourly Aggregations)...")

# Read from Silver
df_silver = spark.read.parquet("s3a://silver/sensor_readings")

# Perform hourly aggregations: calculate average, min, max moisture per sensor
# For test simplicity, group by sensor_id and hour
df_gold = df_silver \
    .withColumn("reading_hour", date_trunc("hour", col("timestamp"))) \
    .groupBy("reading_hour", "sensor_id") \
    .agg(
        avg("reading_value").alias("avg_val"),
        min("reading_value").alias("min_val"),
        max("reading_value").alias("max_val"),
        count("reading_value").alias("reading_count")
    )

# Write to MinIO Gold layer
gold_path = "s3a://gold/sensor_readings_hourly"
print(f"Writing aggregated data to Gold Layer: {gold_path}")
df_gold.write.mode("overwrite").parquet(gold_path)

# Verify Gold contents
df_gold_check = spark.read.parquet(gold_path)
print(f"Gold Layer Record Count: {df_gold_check.count()}")
df_gold_check.show(5, truncate=False)
```

```python
print("Step 6: Loading Gold aggregations into ClickHouse Analytical Database...")

# Let's read from Gold
df_gold_data = spark.read.parquet("s3a://gold/sensor_readings_hourly")

# Map our test data to ClickHouse columns
df_ch_ready = df_gold_data.select(
    col("reading_hour"),
    col("sensor_id"),
    when(col("sensor_id").contains("FIELD-01"), lit(1)).otherwise(lit(2)).alias("field_id"),
    when(col("sensor_id").contains("FIELD-01"), lit("North Field A")).otherwise(lit("South Field B")).alias("field_name"),
    when(col("sensor_id").contains("FIELD-01"), lit("Wheat")).otherwise(lit("Corn")).alias("crop_name"),
    when(col("sensor_id").contains("SM"), col("avg_val").cast("float")).otherwise(lit(45.0).cast("float")).alias("soil_moisture_avg"),
    when(col("sensor_id").contains("SM"), col("min_val").cast("float")).otherwise(lit(45.0).cast("float")).alias("soil_moisture_min"),
    when(col("sensor_id").contains("SM"), col("max_val").cast("float")).otherwise(lit(45.0).cast("float")).alias("soil_moisture_max"),
    when(col("sensor_id").contains("TMP"), col("avg_val").cast("float")).otherwise(lit(22.5).cast("float")).alias("temperature_avg"),
    lit(60.0).cast("float").alias("humidity_avg"),
    col("reading_count").cast("int"),
    when((col("sensor_id").contains("SM")) & (col("avg_val") < 40.0), lit(1).cast("integer")).otherwise(lit(0).cast("integer")).alias("requires_irrigation"),
    lit(0).cast("int").alias("alert_count")
)

# Configure ClickHouse JDBC connection parameters
clickhouse_url = "jdbc:clickhouse://clickhouse:9002/agri_gold"
clickhouse_properties = {
    "user": "agri_analytics",
    "password": "Clickhouse_Secur3_Pass",
    "driver": "com.clickhouse.jdbc.ClickHouseDriver"
}

print("Writing Gold Layer DataFrame to ClickHouse table 'sensor_readings_hourly'...")
df_ch_ready.write \
    .mode("append") \
    .jdbc(url=clickhouse_url, table="sensor_readings_hourly", properties=clickhouse_properties)

print("âœ… ClickHouse write complete!")
```

```python
print("Step 7: Querying ClickHouse database to verify data persistence...")

# Read back table from ClickHouse to verify the JDBC query path
df_ch_read = spark.read \
    .jdbc(url=clickhouse_url, table="sensor_readings_hourly", properties=clickhouse_properties)

print(f"ClickHouse Record Count: {df_ch_read.count()}")
df_ch_read.select("reading_hour", "sensor_id", "field_name", "crop_name", "soil_moisture_avg", "requires_irrigation").show(truncate=False)

print("ðŸŽ‰ CONGRATULATIONS! The entire Smart Agriculture Data Lakehouse integration test is successful!")
```
